```python
ds["Age Group"] = pd.cut(
    ds["Age"],
    bins=[0,20,40,60,80],
    labels=["0-20","21-40","41-60","61-80"]
)

print(pd.crosstab(ds["Age Group"], ds["Gender"]))
```

```python
ax = pd.crosstab(ds["Age Group"], ds["Gender"]).plot(kind="bar")
#crosstab for comparision of 2 column
for container in ax.containers:
    ax.bar_label(container)
#bar_label helps to show value
plt.title("Age and Gender Distribution")
plt.xlabel("Age Group")
plt.ylabel("Gender Count")
plt.show()
```

```python
family_disease = pd.crosstab(ds["Family History"],ds["Heart Disease"])
family_disease
```

```python
ax = family_disease.plot(kind="bar")

for container in ax.containers:
    ax.bar_label(container)

plt.title("Family History vs Heart Disease")
plt.xlabel("Family History")
plt.ylabel("Count")
plt.show()
```

## This analysis compares family history with disease occurrence. It helps identify whether patients with a family history have a higher risk of developing diseases than those without a family history.


```python
corr = ds.corr(numeric_only=True)

print(corr)
```

```python
corr = ds.corr(numeric_only=True)

print(corr["Heart Disease"].sort_values(ascending=False))
```

```python

corr = ds.corr(numeric_only=True)

corr["Heart Disease"].sort_values().plot(kind="bar")

plt.title("Correlation with Heart Disease")
plt.xlabel("Features")
plt.ylabel("Correlation")
plt.xticks(rotation=45)

plt.show()
```

## Correlation analysis measures the relationship between numerical variables. A value close to +1 indicates a strong positive relationship, a value close to -1 indicates a strong negative relationship, and a value near 0 indicates little or no relationship.

```python
high_risk = ds[
    (ds["Age"] > 60) &
    (ds["Blood Pressure"] == "High") &
    (ds["Cholesterol"] == "High")
]

#print(high_risk)
print("Total High-Risk Patients:", len(high_risk))
```

## ## High-risk patient identification is the process of finding patients who have multiple risk factors, such as older age, high blood pressure, high cholesterol, obesity, smoking, or a family history of disease. These patients are more likely to develop serious health conditions and require closer medical attention.
